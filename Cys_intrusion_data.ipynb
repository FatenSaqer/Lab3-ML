```python
plt.figure(figsize=(12,8))
sns.heatmap(df.select_dtypes(include=['number']).corr(), cmap='coolwarm')
plt.title("Correlation Matrix")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x='attack_detected', data=df)
plt.title("Distribution of Attack Detected")
plt.show()
```

```python
plt.figure(figsize=(12,8))

numeric_df = df.select_dtypes(include=['int64','float64'])

sns.heatmap(numeric_df.corr(), cmap='coolwarm', annot=True)
plt.title("Correlation Matrix")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x='protocol_type', hue='attack_detected', data=df)
plt.title("Protocol Type vs Attack")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x='attack_detected', y='failed_logins', data=df)
plt.title("Failed Logins vs Attack Detection")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x='attack_detected', y='session_duration', data=df)
plt.title("Session Duration vs Attack")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x='protocol_type', hue='attack_detected', data=df)
plt.title("Protocol Type vs Attack")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x='attack_detected', y='ip_reputation_score', data=df)
plt.title("IP Reputation Score vs Attack Detection")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x='attack_detected', y='login_attempts', data=df)
plt.title("Login Attempts vs Attack Detection")
plt.show()
```
